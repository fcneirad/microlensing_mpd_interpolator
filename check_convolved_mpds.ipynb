```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
mpd = np.loadtxt('/media/data/imf_with_doubles/convolved/37836.dat')
mpd.shape
```

```python
mpd[:,0] = np.log10(mpd[:,0])
```

```python
disc_sizes = np.linspace(1, 400, 134)/400
```

```python
fig, ax = plt.subplots(1, 1, figsize=(12, 6))
fig.suptitle('$\kappa=0.52 \gamma=0.32 s=0.8$')

ax.plot(mpd[:,0], mpd[:,1], label=str(disc_sizes[0])+'[RE]')
ax.plot(mpd[:,0], mpd[:,26], label=str(disc_sizes[25])+'[RE]')
ax.plot(mpd[:,0], mpd[:,51], label=str(disc_sizes[50])+'[RE]')
ax.plot(mpd[:,0], mpd[:,76], label=str(disc_sizes[75])+'[RE]')
ax.plot(mpd[:,0], mpd[:,101], label=str(disc_sizes[100])+'[RE]')
ax.plot(mpd[:,0], mpd[:,126], label=str(disc_sizes[125])+'[RE]')

ax.set_xlabel('$\mu_{micro} - \mu_{macro}$')
ax.set_ylabel('PDF')
ax.legend()
```
