```python
import numpy as np
import pandas as pd
import sqlite3
```

```python
ids, kappa, gamma, s = np.loadtxt('/mnt/d/data/magnification_maps/masterlist.dat').T
ids = ids.astype(int)

df = pd.DataFrame({'id':pd.Series(ids, dtype=int),
                   'kappa':pd.Series(kappa, dtype=float),
                   'gamma':pd.Series(gamma, dtype=float),
                   's':pd.Series(s, dtype=float)})
```

```python
df[np.logical_and(df['kappa']==0.4, df['gamma']==0.4)]
```

```python
db = sqlite3.connect('/mnt/d/data/magnification_maps/masterlist.db')
df.to_sql('maps', db, if_exists='replace')
```
